## PyTorch Tensors

A **tensor** is a multi-dimensional array, similar to a NumPy `ndarray`, and is the core data structure used in PyTorch. Tensors are used to store the inputs, outputs, and parameters (weights and biases) of a model.

The main advantage tensors have over regular NumPy arrays is that they can run on a GPU or other specialized hardware, which allows much faster computation, especially for large models.

## Converting NumPy Arrays to Tensors

PyTorch tensors can be created directly from existing NumPy arrays using `torch.from_numpy()`. This is useful when data has already been loaded or processed using NumPy, since it avoids having to recreate the data manually.

A tensor created this way shares memory with the original NumPy array, and it keeps the same data type (`dtype`) as the array it came from, unless the type is changed afterward.

## Reshaping and Basic Operations

Tensors can be reshaped into different dimensions using `.reshape()`, as long as the total number of elements stays the same. They also support slicing and indexing similar to NumPy arrays, arithmetic operations (such as squaring values with `.pow()`), and matrix multiplication using functions like `torch.mm()`.

These operations form the basis for how data flows through and is transformed inside a neural network.

**Step 1 — Standard Imports**

```python
import torch
import numpy as np
```

**Step 2 — Define set_seed()**

```python
def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
```

**Step 3 — Create array arr**

```python
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype('int32')

assert np.array_equal(arr, np.array([3,4,2,4,4,1])), "Oops"
print(arr)
```

**Step 4 — Convert arr to tensor x**

```python
x = torch.from_numpy(arr)
print(x)
```

**Step 5 — Change dtype of x**

```python
print('Old:', x.type())

x = x.type(torch.int64)

print('New:', x.type())
```

**Step 6 — Reshape x into 3x2**

```python
x = x.reshape(3,2)
print(x)
```

**Step 7 — Right-hand column of x**

```python
print(x[:,1:])
```

**Step 8 — Squared values of x (unchanged)**

```python
print(x.pow(2))
print(x)
```

**Step 9 — Create tensor y**

```python
set_seed(42)
y = torch.randint(0,5,(2,3))

assert torch.equal(y, torch.tensor([[2, 2, 1],[4, 1, 0]])), "Oops"
print(y)
```

**Step 10 — Matrix product of x and y**

```python
prod = torch.mm(x,y)
prod
```

Since `set_seed(42)` is called before every random draw, both `arr` and `y` produce the exact same values every time this runs, which is why both `assert` checks pass. The tensor `x` starts as a flat 6-element tensor with dtype `int32`, then gets converted to `int64` before being reshaped into a 3x2 matrix. Slicing out the right-hand column doesn't change `x` itself, and squaring it with `.pow(2)` also leaves `x` unchanged since it returns a new tensor instead of modifying it in place. Since `x` has shape (3,2), `y` is created with shape (2,3) so the two can be matrix-multiplied, giving a final 3x3 result.
