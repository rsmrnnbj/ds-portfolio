# Learning Rate and Stability

*How does learning rate change the training curve of a small CNN, and at what point does training become unstable?*

---

## Our Goal

- Explain learning rate in simple words, from the basic idea to how it affects training.
- Use our own experiment to see what happens when we change its value.

## What is learning rate? (hook)

Imagine teaching a computer to recognize clothing images. It makes mistakes, checks how wrong its predictions are, and adjusts itself to improve. But how big should each adjustment be? If the changes are too small, learning might take a long time. If they're too large, the model might struggle to learn properly.

This leads to our question: How does learning rate affect the training of a small CNN?
