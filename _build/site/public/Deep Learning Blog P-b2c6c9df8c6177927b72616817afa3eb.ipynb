# Learning Rate and Stability

How does learning rate affect the training of a small CNN?

---

Imagine teaching a computer to recognize different types of clothing. At first, it will make mistakes. As it learns, it adjusts its values to make better predictions. But how much should it adjust each time?

This is where the learning rate comes in. It controls how big or small the adjustments are during training. If the adjustments are too small, learning can take longer. If they are too large, the model may struggle to learn properly.

In this experiment, we wanted to find out: How does learning rate change the training curve of a small CNN, and where does training become unstable?

To answer this, we'll tested five different learning rates using the Fashion-MNIST dataset.

## 1. Understanding Learning Rate

A neural network learns by adjusting values inside the model called weights and biases. These values help the model recognize patterns and make predictions.

Think of it like walking toward a destination. If your steps are too small, you will take longer to reach it. If your steps are too big, you might go past the destination. The same idea applies to learning rate.

During training, the model follows these steps:
1. It makes a prediction.
2. It checks how wrong the prediction is using a loss function.
3. It calculates a gradient to determine how the model's values should change.
4. It updates its weights and biases to improve its next prediction.

A simplified formula for updating a weight is:

$$w_{\text{new}} = w_{\text{old}} - \alpha \times \text{gradient}$$

Where:

- $w_{\text{old}}$ is the current weight.
- $\alpha$ is the learning rate.
- The gradient indicates how the weight should change.
- $w_{\text{new}}$ is the updated weight.

Here, the learning rate is represented by $\alpha$. It controls how big the adjustment will be.

In this experiment we will be using Adam as the optimizer. An optimizer is the method that updates the model's weights and biases during training. Adam uses information from current and previous gradients to guide its updates. The learning rate is a separate setting that controls the overall size of those updates.

Training is considered more stable when the model improves consistently. For example, when the loss decreases over time, it is usually a good sign. However, when the loss keeps increasing or changing sharply, it may mean that the learning rate is not suitable.

## 2. Experiment Setup

For this experiment, we will used Fashion-MNIST, a dataset containing grayscale images of clothing. Each image is 28 × 28 pixels and belongs to one of 10 clothing categories.

Using a small Convolutional Neural Network (CNN) to classify the images. A CNN is useful for image tasks because it can learn patterns such as shapes, edges, and other visual features.

To see how learning rate affects training, let's test five different values:


| Run | Learning rate |
|---|---:|
| 1 | 0.00001 |
| 2 | 0.0001 |
| 3 | 0.001 |
| 4 | 0.01 |
| 5 | 0.1 |

Only the learning rate changed. The other settings stayed the same so that the comparison would be fair.

This experiment used 10,000 images for training and 6,000 images for validation. Each model was trained for four epochs with a batch size of 64.

We also used three random seeds: 42, 43, and 44. This means each learning rate was tested three times. Doing this helps check whether the results stay similar or change depending on the random starting values.

In total, there were 15 runs: five learning rates tested across three seeds.

## 3. Results and Observations

After running the experiments, we compared the training loss, validation loss, training accuracy, and validation accuracy.

The table below shows the average results from the three runs. The values after ± show how much the results varied across the seeds.

| Learning rate | Final train loss | Final validation loss | Final train accuracy | Final validation accuracy |
|---:|---:|---:|---:|---:|
| 0.00001 | 0.7896 ± 0.0191 | 0.6887 ± 0.0172 | 75.11% ± 0.33% | 78.30% ± 0.42% |
| 0.0001 | 0.3973 ± 0.0022 | 0.3702 ± 0.0021 | 86.12% ± 0.16% | 87.16% ± 0.27% |
| 0.001 | **0.3115 ± 0.0070** | **0.3148 ± 0.0056** | **88.51% ± 0.25%** | **88.47% ± 0.21%** |
| 0.01 | 0.5430 ± 0.0635 | 0.4108 ± 0.0357 | 79.40% ± 2.69% | 84.98% ± 1.21% |
| 0.1 | 1.8402 ± 0.8147 | 1.7783 ± 0.9289 | 28.74% ± 32.73% | 31.17% ± 36.34% |

The results show that 0.001 had the lowest average validation loss and the highest average validation accuracy. The 0.0001 learning rate also performed well, while 0.1 showed the most inconsistent results.

## 4. What Do the Results Tell Us?

The results show that changing the learning rate affected how quickly the model learned and how consistent its performance was.

The smallest learning rate, 0.00001, still improved during training. However, it finished with 78.30% average validation accuracy. This suggests that its updates were too small to reach the same performance as the other better-performing settings within four epochs.

The 0.0001 and 0.001 learning rates gave the best results. Among the five values, 0.001 performed the best, reaching 88.47% average validation accuracy and 0.3148 average validation loss. Its results were also fairly consistent across the three runs.

The 0.01 learning rate also learned, but its average validation accuracy was lower at 84.98%. Its results also varied more than those of 0.0001 and 0.001.

The largest learning rate, 0.1, showed the clearest warning signs. Two of its three runs ended with validation accuracy close to 10%, while the third reached 73.13%. Since Fashion-MNIST has 10 categories, an accuracy close to 10% is similar to guessing.

However, none of the runs produced non-finite loss values such as NaN. So, it would not be correct to say that all three runs completely failed or numerically diverged. Instead, the results show that 0.1 was unreliable for this particular experiment.

Overall, the findings support the idea that a learning rate that is too small can slow down learning, while one that is too large can make training inconsistent.

**Limitations**

There are also some limitations to this experiment. Only 10,000 training images and 6,000 validation images were used instead of the full dataset. The model was also trained for only four epochs, so some learning rates might have performed differently with more training time.

Another limitation is that the experiment used only one CNN architecture and the Adam optimizer. Other models or optimizers may produce different results. Although three random seeds were tested, they are still not enough to make strong statistical claims.

Because of these limitations, the results only show what happened with this particular setup. They do not mean that 0.001 will always be the best learning rate for every CNN.

## 5. Key Takeaways

1. Learning rate matters. It controls how big the model's adjustments are during training. This can affect how quickly the model learns.

2. A bigger learning rate is not always better. In this experiment, 0.1 produced inconsistent results, while 0.001 gave the best average validation performance.

3. It is important to compare multiple runs. Testing different seeds helped show that the same learning rate can produce different results. Looking at only one run may give a misleading impression.
