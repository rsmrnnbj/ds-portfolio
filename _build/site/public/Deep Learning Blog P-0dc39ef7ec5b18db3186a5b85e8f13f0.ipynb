# Learning Rate and Stability

How does learning rate affect the training of a small CNN?

---

## 1. Does the Size of Each Update Matter?

Imagine teaching a computer to recognize clothing images. At first, the model does not know how to identify a shirt, a shoe, or a bag. It makes predictions, checks its mistakes, and adjusts itself to improve.

But how much should it adjust each time?

This is where **learning rate** comes in. Learning rate controls how big the model's adjustments are while learning. If the value is too small, learning may take longer. If it is too large, the model may struggle to improve because its adjustments are too big.

In this post, I will explain what learning rate does and why it matters. I will also test different learning rate values using a small Convolutional Neural Network (CNN) and the Fashion-MNIST dataset.

Question is: **How does learning rate change the training curve of a small CNN, and at what point does training become unstable?**


## 2. Understanding Learning Rate

### What is learning rate?

A neural network learns by adjusting values inside the model called **weights**. These weights help the model recognize patterns and make predictions.

However, the model does not know the correct weights from the beginning. It needs to improve them through training.

One way to understand this is to imagine walking toward a destination. You need to take steps to get closer to it.

- If your steps are too small, you may reach the destination, but it will take longer.
- If your steps are a reasonable size, you can move forward steadily.
- If your steps are too large, you might keep passing the destination instead of reaching it.

Learning rate controls the size of these steps when the model updates its weights.

### How does the model know what to adjust?

During training, the model goes through a few important steps:

1. **Make a prediction.** The model looks at an image and predicts its category.
2. **Measure the mistake.** It compares its prediction with the correct answer using a loss function.
3. **Calculate the gradient.** The gradient helps determine how the weights should change to reduce the mistake.
4. **Update the weights.** The optimizer uses this information to adjust the weights.

This process repeats as the model learns from more examples.

### The basic weight update formula

The effect of learning rate can be shown using this simplified formula:

\[
w_{\text{new}} = w_{\text{old}} - \alpha \times \text{gradient}
\]

Where:

- \(w_{\text{old}}\) is the current weight.
- \(\alpha\) is the learning rate.
- The gradient indicates how the weight should change.
- \(w_{\text{new}}\) is the updated weight.

The learning rate controls how much the gradient affects the update. A larger learning rate means a bigger adjustment, while a smaller one means a smaller adjustment.

### What is an optimizer?

An **optimizer** is the method used to update the model's weights during training. It uses the gradients to help the model improve its predictions.

For this experiment, I will use **Adam**, which stands for Adaptive Moment Estimation. Adam adjusts the updates using information from previous gradients. This helps it adapt the updates for different weights.

Even though Adam has a more detailed update process than the basic formula above, the learning rate still controls the overall size of its updates.

### What is training stability?

Training stability refers to how steadily a model learns during training.

For example, if the model's loss gradually decreases, it suggests that the model is improving. However, if the loss suddenly increases, changes dramatically, or becomes an invalid value, it may indicate that training is unstable.

This is important because choosing a learning rate is not only about helping the model learn faster. It is also about finding a value that allows it to learn properly.

The task of the model is to identify the correct category for each image.
