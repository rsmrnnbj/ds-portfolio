# 04 — Data Cleaning

THIS should definitely be its own category.

Inspect → Identify problems → Clean → Check again

**How to check for missing values**

```python
df.isnull().sum()
```

**How to remove rows with missing values**

```python
df.dropna()
```

**How to fill missing values**

```python
df["Age"] = df["Age"].filledna(df["Age"].mean())
```

**How to check for duplicates**

```python
df.duplicated().sum()
```

**How to remove duplicates**

```python
df = df.drop_duplicates()
```

**How to check for inconsistent text**

```python
df["Category"].value_counts()
```

This can help reveal values such as:

- Male
- male
- MALE

**How to standardaize text**

```python
df["Category"] = df["Category"].str.lower()
```

**How to remove extra spaces**

```python
df["Name"] = df["Name"].str.strip()
```

**How to change a column's data type**

```python
df["Age"] = df["Age"].astype(int)
```

**How to convert a column to datetime**

```python
df["Date"] = pd.to_datetime(df["Date"])
```
